# Cat Flag

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(0)
df=pd.DataFrame(np.random.randint(low=50,high=500,size=50),columns=["org_val"])
df.head(2)
```

## 1 apply

```python
df["$"]=df["org_val"].apply(lambda x: "${:.0f}".format(x))
df.head(2)
```

## 2

```python
df["High_Low"]=np.where( ~df["org_val"].between(60,490) ,"Outliers" ,"Normal_values")
df[df["High_Low"] == "Outliers"]
```

```python
df.query("High_Low == 'Outliers'")
```

## >2

```python
df['bins']=pd.cut( df['org_val'],
                   bins = [50,100,150,200,250,300,350,400,450,500],
                   labels=list(map(lambda x : "Level_{}".format(chr(x)) , np.arange(65,65+9))))
df.head(2)
```

## mapper

```python
def mapper(x):
    if x <100 :
        return "BAD"
    elif x >=100 and x<200:
        return "Good"
    elif x >=200 and x<300:
        return "Perfect"
    elif x >=300 and x<400:
        return "Super"
    elif x >=400 and x<500:
        return "Awesome"
    else:
        return "Jackpot"

df['map']=df['org_val'].map(mapper)
df.query("map in ['BAD','Good']").head(2)
```

```python
dict_mapper={"BAD":"Unsatisfying",
             "Good":"Improvement_Expected"}
df["map_new"]=df['map'].map(dict_mapper)
df.head(5)
```
